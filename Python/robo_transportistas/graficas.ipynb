## Gráfica de barras - Tipo de incidente

```python
incidente = pd.read_excel('base_informacion.xlsx', sheet_name='slide1_barras')
print(incidente.shape); incidente.head(5)
```

```python
incidente['etiqueta'] = incidente['etiqueta'].str.replace(r' ', r'\n', regex=True)
nombres_incidente = incidente['etiqueta'].tolist()
nombres_incidente
```

```python
incidente = incidente.drop(columns=['etiqueta'])
print(incidente.shape); incidente.head(5)
```

```python
incidente = incidente[['id', 'extorcion', 'secuestro de unidad', 'apedreamiento', 'robo']]
incidente = incidente.rename(columns={
    'id': 'ID',
    'extorcion': 'Extorsión',
    'secuestro de unidad': 'Secuestro de unidad',
    'apedreamiento': 'Apedreamiento',
    'robo': 'Robo'
})
print(incidente.shape); incidente.head(5)
```

```python
barras_verticales(incidente, 
       nombre="Gráfica de barras - tipo de incidente", 
       ancho_barra=0.9, 
       tipo_letra='Montserrat',
       valor_barra=True, 
       tam_letra_valor_barra=20,
       porce_barra=False, 
       porce_total=False,
       ordenar_por='etiqueta',
       orden='ascendente',
       area_min=0.13,
       leyenda=False,
       sustituir_etiquetas=nombres_incidente,
       tam_letra_valor_capsu=26,
       paleta_colores=['#898989', '#a7802d', '#621132', '#124741'],
       quitar_capsu=True,
       ajusta_pos_capsu=0.00,
       color_valor_capsu='black',
       color_valor_barra='white',
       orientacion_etiqu_ejeX='horizontal',
       tam_letra_ejeX=18,
       color_letra_ejeX='#7c7c7c',
       tam_letra_ejeY=25,
       color_letra_ejeY='#7c7c7c',
       fondo='transparent',
       mostrar_fig=False,
       guardar_fig=True,
       #decimales=1,
)
```
