```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd

# Crear el DataFrame con años
df = pd.DataFrame({
    'Año': [2019, 2020, 2021, 2022],
    'Positivo 1': [5, 3, 4, 2],
    'Positivo 2': [2, 1, 3, 1],
    'Negativo 1': [-2, -4, -1, -3],
    'Negativo 2': [-1, -2, -2, -1]
})
df
```

```python
x = np.arange(len(df['Año']))

fig, ax = plt.subplots(figsize=(8, 5))

# Barras apiladas positivas (arriba del eje)
ax.bar(x, df['Positivo 1'], color='#4CAF50', label='Positivo 1')
ax.bar(x, df['Positivo 2'], bottom=df['Positivo 1'], color='#81C784', label='Positivo 2')

# Barras apiladas negativas (abajo del eje)
ax.bar(x, df['Negativo 1'], color='#F44336', label='Negativo 1')
ax.bar(x, df['Negativo 2'], bottom=df['Negativo 1'], color='#FF8A65', label='Negativo 2')

ax.axhline(0, color='black', linewidth=1)
ax.set_xticks(x)
ax.set_xticklabels(df['Año'])
ax.set_ylabel('Valor')
ax.set_title('Diverging Stacked Bar Chart por Año')
ax.legend(loc='upper left')
plt.show()
```
