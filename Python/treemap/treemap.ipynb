### Dataframe de ejemplo para el treemap

```python
import pandas as pd

victimas_por_entidad = pd.DataFrame({
    'ENTIDAD FEDERATIVA': [
        'Jalisco', 'Guanajuato', 'Michoacán', 'Sonora', 'Oaxaca',
        'Estado de México Estado de México Estado de México Estado de México Estado de México Estado de México', 'Durango', 'Nuevo León', 'Coahuila', 'Tamaulipas',
        'Puebla', 'Ciudad de México', 'Guerrero', 'Veracruz', 'Chiapas'
    ],
    'NUMERO DE VICTIMAS': [
        910, 750, 664, 516, 508,
        485, 380, 320, 264, 171,
        156, 152, 137, 121, 70
    ]
})
print(victimas_por_entidad.shape); victimas_por_entidad.head(5)
```

```python
treemap(victimas_por_entidad, 
        nombre="victimas_por_entidad",
        porce_parentesis=True,
        paleta_colores=['#ff9999','#66b3ff','#99ff99','#ffcc99','#c2c2f0','#ffb3e6'],
        )
```

```python
camiones = pd.DataFrame({
    'ENTIDAD FEDERATIVA': [
        'Tractocamión Articulado', 'Autobús', 'Camión Unitario', 'Tractocamión Doblemente Articulado', 'Tractocamión',
        'C-R'
    ],
    'NUMERO DE VICTIMAS': [
        700, 572, 226, 69, 7,
        3
    ]
})
print(camiones.shape); camiones.head(5)
```
