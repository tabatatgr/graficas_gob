## Líneas multiples

```python
# Generar datos
fechas_2 = pd.date_range(start="2000-01-01", end="2025-12-31", freq="YE")
n_2 = len(fechas_2)

# Generar valores aleatorios que suben con el tiempo y caen bruscamente al final
valores_2 = {
    "Fecha": fechas_2,
    "Desaparecidos": np.concatenate([
        np.sort(np.random.uniform(100, 4000, n_2 - 5)),  # Valores ascendentes aleatorios
        np.random.uniform(500, 1000, 5)  # Caída brusca al final
    ]),
    "Localizados": np.concatenate([
        np.sort(np.random.uniform(200, 3500, n_2 - 5)),
        np.random.uniform(300, 800, 5)
    ])
}

# Crear DataFrame
df_2 = pd.DataFrame(valores_2)

# Transformar el DataFrame a formato long
df_2 = pd.melt(
    df_2,
    id_vars=["Fecha"],
    value_vars=["Desaparecidos", "Localizados"],
    var_name="Variable",
    value_name="Valor"
)

df_2
```

```python
lineas(
    df=df_2,
    nombre='lineas_multiples',
    tam_letra_ejeX=20,
    orientacion_etiqu_ejeX='vertical',
    )
```

## Escalonadas

```python
fechas_3 = pd.date_range(start='2024-01-01', end='2025-07-01', freq='MS')

valores_3 = [
    2.1, 0.8, 1.7, 2.5, 1.2, 2.9, 0.6, 1.5, 2.3, 1.1, 2.7, 1.9, 0.9, 2.6, 1.4,
    2.2, 0.7, 2.8, 1.3
]  # 19 valores para 19 fechas

df_3 = pd.DataFrame({
    'FECHA': fechas_3,
    'CATEGORIA': ['Promedio diario'] * len(fechas_3),
    'VALOR': valores_3
})

print(df_3.shape); df_3.head(5)
```

```python
lineas(df_3,
       nombre='linea', 
       tipo_letra='Montserrat',
       ancho_fig=6,
       alto_fig=4.5,
       tam_letra_ejeX=7.4,
       tam_letra_ejeY=16,
       color_letra_ejeX='#535353',
       color_letra_ejeY='#535353',
       weight_letra_ejeX='bold',
       weight_letra_ejeY='bold',
       escalonada=True,
       paleta_colores=['#691c32'],
       modalidad_fechas=2,
       lineas_verticales=[
           {'x': pd.to_datetime('2024-10-01'), 'color': '#DC8F12', 'linewidth': 3},
       ],
       sin_tag=100000,
       tag_max=True,
       tag_final=True,
       tag_especifico=[pd.to_datetime('2024-10-01') - pd.DateOffset(months=1)],
       decimales_capsu=1,
       rounding_size=0.5,
       capsula_valor_weight='bold',
       ancho_linea=3.5,
       margen=True,
       margen_color='#e5e5e5',
       margen_ancho=3,
       marcas_ejes=False,
       mostrar_fig=False,
       guardar_fig=False,
       ajusta_pos_capsu=0.07,
       reducir_ticks_y=True, 
       )
```
